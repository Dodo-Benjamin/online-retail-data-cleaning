```python

# Examinons les lignes dont la quantité est négative
df_negatif = df[df["Quantity"] < 0]

print("Nombre de lignes avec quantité négative :",
      len(df_negatif))

# Examinons quelques factures concernées
print(
    df_negatif[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(20).to_string(index=False)
)
```

```python

# Identifions les factures dont le numéro commence par C
annulations = df[
    df["InvoiceNo"].str.startswith("C", na=False)
]

# Identifions les lignes à quantité négative
quantites_negatives = df[df["Quantity"] < 0]

print("Lignes de factures commençant par C :",
      len(annulations))

print("Lignes avec quantité négative :",
      len(quantites_negatives))

print("Quantités négatives sans préfixe C :",
      len(quantites_negatives[
          ~quantites_negatives["InvoiceNo"].str.startswith(
              "C", na=False
          )
      ]))
```

```python

# Sélectionnons les quantités négatives sans préfixe C
negatives_sans_C = df[
    (df["Quantity"] < 0)
    & (~df["InvoiceNo"].str.startswith("C", na=False))
]

# Affichons les 20 premières lignes
print(
    negatives_sans_C[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID"]
    ].head(20).to_string(index=False)
)
```

```python

# Identifier les lignes qui cumulent les 3 caractéristiques
anomalies = df[
    (df["Quantity"] < 0)
    & (df["UnitPrice"] == 0)
    & (df["Description"].isna())
]

print("Nombre de lignes concernées :", len(anomalies))

print("\nRépartition des quantités :")
print(anomalies["Quantity"].describe())

print("\nExemples de codes produits :")
print(anomalies["StockCode"].value_counts().head(10))
```

```python

# Afficher les 10 lignes ayant les quantités les plus négatives
colonnes = [
    "InvoiceNo",
    "StockCode",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country"
]

print(
    anomalies.nsmallest(10, "Quantity")[colonnes]
    .to_string(index=False)
)
```

```python

# Calculer le pourcentage de valeurs manquantes
pourcentage_manquant = (
    df.isnull().mean() * 100
).round(2)

print(pourcentage_manquant)
```

```python

import pandas as pd
import numpy as np

# Conserver les données originales et travailler sur une copie
df_clean = df.copy()

# Vérifier la structure du DataFrame
print("Dimensions :", df_clean.shape)

print("\nTypes de données :")
print(df_clean.dtypes)

print("\nValeurs manquantes :")
print(df_clean.isna().sum())

print("\nDoublons exacts :")
print(df_clean.duplicated().sum())
```

```python

# Examiner les descriptions manquantes
description_manquante = df_clean[
    df_clean["Description"].isna()
]

print("Nombre de descriptions manquantes :",
      len(description_manquante))

print("\nExemples de lignes concernées :")
display(description_manquante.head(10))

print("\nNombre de codes produits distincts concernés :",
      description_manquante["StockCode"].nunique())
```

```python

# Rechercher les descriptions connues pour chaque code produit
descriptions_connues = (
    df_clean.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

# Identifier les lignes dont la description est manquante
masque = df_clean["Description"].isna()

# Récupérer les descriptions uniquement si le code possède
# une description connue ailleurs dans le fichier
df_clean.loc[masque, "Description"] = (
    df_clean.loc[masque, "StockCode"]
    .map(descriptions_connues)
)

print("Descriptions manquantes après vérification :",
      df_clean["Description"].isna().sum())

print("\nDescriptions récupérées :",
      masque.sum() - df_clean["Description"].isna().sum())
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "data/archive/data.csv",
    encoding="ISO-8859-1"
)

print("Dimensions :", df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.head())
```

```python
df_clean = df.copy()

print("Dimensions de df_clean :"
, df_clean.shape)
```

```python
print("Dimensions :", df_clean.shape)

print("\nDoublons exacts :", df_clean.duplicated().sum())

print("\nValeurs manquantes :")
print(df_clean.isna().sum())

print("\nTypes de données :")
print(df_clean.dtypes)
```

```python
# Suppression des doublons exacts
df_clean = df.copy()

print("Avant suppression :", df_clean.shape)

df_clean = df_clean.drop_duplicates()

print("Après suppression :", df_clean.shape)

print("Doublons restants :", df_clean.duplicated().sum())
```

```python
# Conversion de InvoiceDate en date/heure
df_clean["InvoiceDate"] = pd.to_datetime(
    df_clean["InvoiceDate"],
    errors="coerce"
)

print(df_clean.dtypes)
```

```python
# Verification des dates impossibles à convertir
print(
    "Dates non converties :",
    df_clean["InvoiceDate"].isna().sum()
)
```

```python
# Construire une correspondance StockCode → Description
descriptions_connues = (
    df_clean.dropna(subset=["Description"])
    .groupby("StockCode")["Description"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

# Identifier les descriptions manquantes
masque_description = df_clean["Description"].isna()

# Récupérer les descriptions connues
df_clean.loc[masque_description, "Description"] = (
    df_clean.loc[masque_description, "StockCode"]
    .map(descriptions_connues)
)

print(
    "Descriptions manquantes après traitement :",
    df_clean["Description"].isna().sum()
)
```

```python
# Conversion de CustomerID en entier nullable
df_clean["CustomerID"] = df_clean["CustomerID"].astype("Int64")

print(df_clean.dtypes)
```

```python
print(
    "CustomerID manquants :",
    df_clean["CustomerID"].isna().sum()
)
```

```python
# Nettoyage des espaces dans Description
df_clean["Description"] = (
    df_clean["Description"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
# Nettoyage des espaces dans Country
df_clean["Country"] = (
    df_clean["Country"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

```python
print("Valeurs uniques dans Country :")
print(df_clean["Country"].unique())

print("\nNombre de descriptions manquantes :",
      df_clean["Description"].isna().sum())
```

```python
print("========== VALIDATION FINALE ==========")

print("\nDimensions :")
print(df_clean.shape)

print("\nDoublons exacts :")
print(df_clean.duplicated().sum())

print("\nTypes de données :")
print(df_clean.dtypes)

print("\nValeurs manquantes :")
print(df_clean.isna().sum())

print("\nQuantités négatives :")
print((df_clean["Quantity"] < 0).sum())

print("\nPrix unitaires négatifs :")
print((df_clean["UnitPrice"] < 0).sum())
```
